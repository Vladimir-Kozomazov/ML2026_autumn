```python
import numpy as np # Для работы с многомерными массивами
import matplotlib.pyplot as plt # Визуализация
import matplotlib.image # Чтение изображения
```

```python
img = matplotlib.image.imread('ЧБ.jpg.avif') # Чтение изображения (трёхмерный массив)
R = img[:,:,0] # Красный канал (Все строки, все столбцы, красный цвет)
G = img[:,:,1] # Зеленый канал
B = img[:,:,2] # Синий канал
img_gray = 0.299*R + 0.587*G + 0.114*B # Превращаем изображение в ЧБ (Взвешенная суммка красной, синей и зелёной составляющей)
h, w = img_gray.shape # Берём высоту и ширину нашей картинки (число строк и столбцов)
original_flat = img_gray.flatten() # Многомерный массив превращаем в одномерный, получаем вектор h * w

plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.show()
print(h,w)
```

Часть А: Исходное изображение — это набор пикселей, и в таком виде с ним неудобно проводить вычисления. Поэтому мы переводим картинку в числовой массив, с которым можно работать математически. Изображение цветное и состоит из трёх каналов (R, G, B), то есть хранит в три раза больше чисел, чем нужно для базового анализа. Чтобы уменьшить объём данных, мы переводим картинку в оттенки серого по стандартной формуле — в результате получается одна матрица, полностью описывающая изображение.

```python
U, s, Vt = np.linalg.svd(img_gray, full_matrices=False) # svd (Расскладываем матрицу на 3 матрицы), False - урезание матрицы
list_of_ranks = [1, 5, 10, 30, 100, min(img_gray.shape)] # Сколько сингулярных чисел мы оставим, остальные обнулим - ранг; min(img_gray.shape) - полный ранг
original_shape = (h,w) # Создали кортеж чисел, чтобы могли сравнивать с оригиналом наши сжатые изображения

fig, axes = plt.subplots(2, 3, figsize=(12, 8)) # Готовим места под графики

approximations = [] # Список для восстановленных матриц
for i,r in enumerate(list_of_ranks):
    Ar = (U[:, :r] * s[:r]) @ Vt[:r, :] # Восстановление для ранга r (Каждый столбец умножается на своё сингулярное число) * (Матрицу r * w)
    approximations.append(Ar) # Кладём восстановленную матрицу в список

    axes.flat[i].imshow(Ar, cmap='gray') # Вывод картинки в ячейку
    axes.flat[i].set_title(f'Ранг = {r}\nСжатие в {((3*w*h))/(r*(h+1+w)*3*4)} раз')
    axes.flat[i].axis('off')

plt.tight_layout()
plt.show()
```

Часть Б: Изображение раскладывается на три матрицы, и от каждой из них мы берём только первые r чисел. Чем меньше ранг, тем сильнее теряется различимость картинки. При r = 30 изображение приобретает лишь примерные очертания. При r = 100 объект уже легко узнаётся, но с заметным шумом. А при r = 1080 картинка восстанавливается практически без потерь.

Часть В: Для наглядного сравнения все восстановленные изображения выводятся на одном графике в сетке 2×3.
Часть Г: Мы сравниваем исходное изображение со сжатым и считаем разницу между ними. Чем меньше ранг, тем выше коэффициент сжатия — и тем меньше весит итоговая картинка.

```python
# ---------- БЛОК АВТОПРОВЕРКИ (НЕ РЕДАКТИРОВАТЬ) ----------
# Предполагается, что у вас есть переменные:
# original_shape = (h, w, c) или (h, w)
# list_of_ranks = [1, 5, 10, 30, 100, ...]
# approximations = список восстановленных массивов (numpy) для каждого ранга

# 1. Проверка размерностей
for i, r in enumerate(list_of_ranks):
    assert approximations[i].shape == original_shape, f"Ошибка: размерность для ранга {r} не совпадает с исходной"

# 2. Проверка, что значения не выходят за пределы 0-255 (для RGB/серого)
for i, r in enumerate(list_of_ranks):
    arr = approximations[i]
    if arr.dtype != np.uint8:
        arr = np.clip(arr, 0, 255) # Если float, то проверяем диапазон
    assert arr.max() <= 255.1, f"Значения превышают 255 для ранга {r}"
    assert arr.min() >= -0.1, f"Значения меньше 0 для ранга {r}"

# 3. Проверка, что с ростом ранга ошибка уменьшается (метрика MSE)
from sklearn.metrics import mean_squared_error
mse_list = []
for i in range(len(approximations)):
    # Если цветная, считаем MSE по всем каналам
    mse = mean_squared_error(original_flat, approximations[i].flatten())
    mse_list.append(mse)

# Проверяем монотонность (MSE должен падать, так как мы добавляем сингулярные числа)
for i in range(1, len(mse_list)):
    assert mse_list[i] <= mse_list[i-1] + 1e-6, f"MSE не уменьшается между рангами {list_of_ranks[i-1]} и {list_of_ranks[i]}"

print("✅ Все автоматические проверки пройдены. Задание выполнено корректно!")
```

В ходе работы была решена задача сжатия растрового изображения с контролируемой потерей качества. Перебрав несколько значений ранга, мы убедились, что визуально картинка перестаёт отличаться от оригинала примерно при r = 100. Чем меньше ранг, тем сильнее сжатие. Чтобы получить сжатие в 10 раз, достаточно ранга r ≈ 17, однако при таком коэффициенте изображение хоть и различимо, но теряет в качестве заметно. Ошибка воспринимается по-разному на тёмных и светлых участках: у тёмных пикселей значения малы, поэтому погрешность составляет большую долю от их яркости, тогда как на светлых участках та же абсолютная ошибка выглядит незначительной и почти не заметна.
